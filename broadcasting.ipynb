                    np.hstack() ==> side to side add

```python
#np.vstack() is used to stack arrays in sequence vertically (row wise). It takes a tuple of arrays as input and returns a single array that is the vertical concatenation of the input arrays. The input arrays must have the same number of columns.
arr1 = np.array([[1, 2, 3], [4, 5, 6]])
arr2 = np.array([[7, 8, 9], [10, 11, 12]])

arr3 = np.vstack((arr1,arr2)) # vertical stack
print("Vertical stack of arr1 and arr2:\n", arr3)
```

```python
#np.hstack() is used to stack arrays in sequence horizontally (column wise). It takes a tuple of arrays as input and returns a single array that is the horizontal concatenation of the input arrays. The input arrays must have the same number of rows.
arr4 = np.hstack((arr1, arr2)) # horizontal stack
print("Horizontal stack of arr1 and arr2:\n", arr4)
```

```python
# np.concatenate() is used to join two or more arrays along an existing axis. It takes a tuple of arrays as input and returns a single array that is the concatenation of the input arrays. The input arrays must have the same shape, except in the dimension corresponding to the axis along which they are being concatenated.
# using axis=0, the arrays are concatenated vertically (row wise), while using axis=1, the arrays are concatenated horizontally (column wise).

print(arr3, arr4)
print("Concatenation of arr3 and arr4 along axis 0:\n", np.concatenate((arr1, arr2), axis=0))
```
