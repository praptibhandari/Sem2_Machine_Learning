## 7. Random Module

_`np.random.seed(n)` → reproducibility | `.rand()` → uniform | `.randn()` → normal | `.randint(low,high,size)` → random ints | `.choice(arr,size)` → random sample | `.shuffle(arr)` → shuffle in place_


**Q7.1** Set a seed, generate a 3x3 uniform random array, and confirm reproducibility by rerunning with the same seed.

```python
# Write your code here
np.random.seed(42)

arr1 = np.random.rand(3, 3)

np.random.seed(42)

arr2 = np.random.rand(3, 3)

print("First array:")
print(arr1)

print("\nSecond array:")
print(arr2)

print("\nSame:", np.array_equal(arr1, arr2))
```

## 11. Flatten vs Ravel

_`.flatten()` → returns a copy (original safe) | `.ravel()` → returns a view (original can change too)_


**Q11.1** Create a 2D array, flatten it with both methods, modify the first element of each result, and check which one affects the original array.

```python
# Write your code here
arr = np.array([
    [1, 2],
    [3, 4]
])

flat = arr.flatten()
ravel = arr.ravel()

flat[0] = 100
ravel[0] = 200

print("Original:")
print(arr)

print("Flatten:")
print(flat)

print("Ravel:")
print(ravel)
```

## 12. Stacking & Concatenation

_`np.vstack()` → stack rows | `np.hstack()` → stack columns | `np.concatenate(axis=)` → general version_


**Q12.1** Create two 2x2 arrays. Combine them using vstack, hstack, and concatenate on both axis=0 and axis=1.

```python
# Write your code here
a = np.array([
    [1, 2],
    [3, 4]
])

b = np.array([
    [5, 6],
    [7, 8]
])

print("vstack:")
print(np.vstack((a, b)))

print("\nhstack:")
print(np.hstack((a, b)))

print("\nconcatenate axis=0:")
print(np.concatenate((a, b), axis=0))

print("\nconcatenate axis=1:")
print(np.concatenate((a, b), axis=1))
```

## 14. Matrix Operations & Transpose

_`*` → element-wise multiply | `@` / `np.dot()` → matrix multiplication | `.T` → transpose (rows ↔ cols)_


**Q14.1** Create two 2x2 matrices. Compute their element-wise product and their matrix product (using both @ and np.dot).

```python
# Write your code here
a = np.array([
    [1, 2],
    [3, 4]
])

b = np.array([
    [5, 6],
    [7, 8]
])

print("Element-wise product:")
print(a * b)

print("\nMatrix product using @:")
print(a @ b)

print("\nMatrix product using np.dot:")
print(np.dot(a, b))
```

**Q14.2** Print the transpose of the first matrix.

```python
# Write your code here
print(a.T)
```

**Q18.1** Create a 5x5 array (1-25), extract its center 3x3 block, then find the row-wise sum of that block.

```python
# Write your code here
arr = np.arange(1, 26).reshape(5, 5)

print("Original:")
print(arr)

center = arr[1:4, 1:4]

print("\nCenter 3x3:")
print(center)

row_sum = center.sum(axis=1)

print("\nRow-wise sum:")
print(row_sum)
```

**Q18.2** Create two 1D int arrays, stack them with vstack, add 0.5 to the result and check the dtype coercion.

```python
# Write your code here
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

result = np.vstack((a, b))

print("Before adding 0.5:")
print(result)
print("dtype:", result.dtype)

result = result + 0.5

print("\nAfter adding 0.5:")
print(result)
print("dtype:", result.dtype)
```
