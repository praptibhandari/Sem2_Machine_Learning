```python
def clean(name):
    name = name.strip().title()
    def check(name):
 
        name = name.replace(" ","")
        if name.isalpha():
            return name
        
    if check(name):
        print(name)
    else:
        print("Name contains invalid characters")

clean(" rahul KUMAR sharma ")
```

```python
marks = "45,88,72,91,60"
lst = list(map(int, marks.split(',')))
total=0
for i in lst:
    total+=i
    average = total/len(lst)

print(total, average) 
```
