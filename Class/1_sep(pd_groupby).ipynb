```python
import pandas as pd
tips = pd.read_csv('../Data/tip.csv')
booking = pd.read_csv('../Data/bookings.csv')
```

```python
# bill count
a = tips.groupby("day")["total_bill"].size()
print(a)

# avg bill
b = tips.groupby("day")["total_bill"].mean()
print(b)

#tip avg
c = tips.groupby("day")["tip"].mean()
print(c)


# all together
# d = tips.groupby("day").agg({"total_bill": ["size", "mean"], "tip": "mean"})
# print(d)


tips.groupby("day").agg(
    Bill_count = ("total_bill", "size"),
    Bill_avg =("total_bill", "mean"),
    Bill_tip = ("tip", "mean")
)   
```

```python
tips.groupby("time").agg(
    Bill_count = ("total_bill", "size"),
    Bill_avg =("total_bill", "mean"),
    Avg_party = ("size", "mean")
)
```

```python
# only good days
# the day where the avg tip is under 3

summary = tips.groupby("day").agg(
    Bill_count = ("total_bill", "size"),
    Bill_avg =("total_bill", "mean"),
    Avg_tip = ("tip", "mean")
).round(2)

# the round function rounds the values to 2 decimal places
print(summary)

summary[summary["Avg_tip"] <= 3]

# summary[summary["day"] == "Sun"] gives error cuz the day column is not in the summary dataframe. It is an index. So we need to use .loc to access it.
# the column we use to groupby becomes the index of the resulting dataframe. So we need to use .loc to access it.

```

```python
# making the index a column again
print(summary.columns.to_list())

print(summary.index.to_list())


# reset_index() makes the index a column again. It also creates a new index for the dataframe. 
# The new index is a range index starting from 0. The old index is added as a new column to the dataframe.
#  The name of the new column is the name of the old index. If the old index has no name, the new column will be named 'index'. 
# If the old index has a name, the new column will be named after the old index's name.
flat = summary.reset_index()
print(flat.columns.to_list())
flat.head()
```

```python
# email me the file.
# wants summary to_csv() and send it to me.

summary.to_csv("summary.csv")
flat= pd.read_csv("summary.csv")
print(flat.head())
```

```python
# somking-section report.
# per smoker grp, the no of bills, avg bill, avg tip. save the report as a csv file, so that YES/NO lable survives

smoker_summary = tips.groupby("smoker").agg(
    Bill_count = ("total_bill", "size"),
    Bill_avg =("total_bill", "mean"),
    Avg_tip = ("tip", "mean")
).round(2)

# reset indeex
flat = smoker_summary.reset_index()
print(flat.columns.to_list())
flat.head() 

# converting into csv file
smoker_summary.to_csv("smoker_summary.csv")

# printing the csv file
flat= pd.read_csv("smoker_summary.csv")
print(flat.head())
```
