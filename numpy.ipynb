```python
# 1. Create a 1D array of integers from 1 to 10.
a = np.arange(1,11)
print("list:", a)


# 2. Creata a 3X3 martix of all zeros, and another of all ones.
mat_1 = np.zeros((3,3))
mat_2 = np.ones((3,3))
print("Matrix of zeros:")
print(mat_1)
print("Matrix of ones:")
print(mat_2)

# 3.Create an array of all even number b/w 2 and 20(inclusive) using np.arange().
nums = np.arange(2,21,2)
print("Even numbers:", nums)

# 4.Crete an array of 5 part from 0 to 1 using np.linspace().
arr = np.linspace(0,1,5)
print("Linspace array:", arr)
```

```python
mat = np.arange(5*4).reshape(5,4)
print("matrix:")
print(mat)

print("Rows:")
print(mat[::4,::3])

```

```python
rolls = np.random.randint(35,101,20)
print(np.mean(rolls))
print(np.median(rolls))
print(np.std(rolls))

```

```python
# min, max, sum, mean, median, std, var
arr = np.arange(6).reshape(2,3)
print(arr)
print("min", np.min(arr))
print("row min", arr.min(axis=1))
print("col min", arr.min(axis=0))

print("sum of the array:", np.sum(arr))
print("sum of the rows:", arr.sum(axis=1))
print("sum of the columns:", arr.sum(axis=0))

print("meadian", np.median(arr))
print("std", np.std(arr))
```
