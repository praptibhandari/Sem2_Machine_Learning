```python
#8
X = np.array([[50, 1200, 3],
              [80, 1500, 9],
              [65, 900, 5],
              [95, 2000, 7]])

col_min = X.min(axis=0)
col_max = X.max(axis=0)

normalized = (X - col_min) / (col_max - col_min)

print(normalized)
print("Column minimums:", normalized.min(axis=0))
print("Column maximums:", normalized.max(axis=0))
```

```python
#13
A = np.array([[1, 2, 3],
              [4, 5, 6]])

B = np.array([[7, 8, 9],
              [1, 0, 2]])

result1 = A @ B.T
result2 = A.T @ B

print(result1)
print(result1.shape)

print(result2)
print(result2.shape)

# A @ B is invalid because:
# A has shape (2,3) and B has shape (2,3).
# For matrix multiplication, inner dimensions must match.
```

```python
#14
hours = np.array([2, 8, 5, 10, 3, 7])
attendance = np.array([55, 92, 70, 98, 60, 85])

# Convert into column vectors
hours_col = hours.reshape(-1, 1)
attendance_col = attendance.reshape(-1, 1)

print("hours_col:")
print(hours_col)
print("Shape:", hours_col.shape)

print("attendance_col:")
print(attendance_col)
print("Shape:", attendance_col.shape)


# Join into two-column table
table = np.hstack((hours_col, attendance_col))

print("Table:")
print(table)
print("Shape:", table.shape)


# Calculate total for each row
total = table.sum(axis=1).reshape(-1, 1)

print("Total:")
print(total)
print("Shape:", total.shape)


# Attach total as third column
three_column = np.hstack((table, total))

print("Three-column table:")
print(three_column)
print("Shape:", three_column.shape)


# Weighted score
w = np.array([0.5, 0.2, 0.3])

scores = three_column @ w

print("Scores:")
print(scores)
print("Shape:", scores.shape)


# Average
average = scores.mean()

print("Average:", average)


# Above average = 1, otherwise 0
result = np.where(scores > average, 1, 0)

print("Final result:", result)
print("Shape:", result.shape)
```

```python
#16 flatten() v/s ravel()
m = np.array([[4, 8, 15],
              [16, 23, 42]])

flat = m.flatten()
rav = m.ravel()

flat[0] = 999
rav[0] = 888

print("Flatten:", flat)
print("Ravel:", rav)

print("Original matrix:")
print(m)

# flatten() creates a copy, so changing flat does not affect m.
# ravel() usually returns a view, so changing rav changes m.
```

```python
#17 Temperature anayalsis
np.random.seed(7)

temps = np.random.randint(15, 41, 28)

weeks = temps.reshape(4, 7)

weekly_average = weeks.mean(axis=1)

hottest_week = np.argmax(weekly_average)

days_above_35 = np.sum(temps > 35)

print("Weeks:")
print(weeks)

print("Weekly averages:")
print(weekly_average)

print("Hottest week index:", hottest_week)
print("Days above 35:", days_above_35)
```

```python
#18. Is matrix multiplication commutative?
np.random.seed(10)

A = np.random.randint(1, 10, (3, 3))
B = np.random.randint(1, 10, (3, 3))

AB = A @ B
BA = B @ A

print("A:")
print(A)

print("B:")
print(B)

print("A @ B:")
print(AB)

print("B @ A:")
print(BA)

print("Are they equal?", np.array_equal(AB, BA))

# Matrix multiplication is generally NOT commutative.
# Therefore, A @ B is generally different from B @ A.
```

```python
#19 Replace marks below 35

marks = np.array([[45, 28, 67, 80],
                  [33, 90, 55, 41],
                  [72, 66, 30, 58],
                  [88, 79, 91, 84],
                  [25, 40, 38, 52]])

marks[marks < 35] = 35

totals = marks.sum(axis=1)

topper = np.argmax(totals)

print(marks)
print("Student totals:", totals)
print("Topper index:", topper)
print("Topper total:", totals[topper])
```
