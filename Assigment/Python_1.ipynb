Question 1: The Strict Username Validator
Task: Write a function that takes a username string as input and validates it. To be valid, the username must be at least 5 characters long after removing any leading/trailing spaces, and it must contain only letters and numbers (no special characters like @, _, or spaces inside). If it is valid, return the clean, strictly lowercase version of the username. If it is invalid, return the string "Invalid".

Input 1:

" CodeMaster99 "


Input 2:

" Python_Dev "


Expected Output 1:

"codemaster99"


Expected Output 2:

"Invalid"

```python
def validate_username(username):
    username = username.strip()

    if len(username) < 5:
        return "Invalid"

    if not username.isalnum():
        return "Invalid"

    return username.lower()

a = validate_username(" CodeMaster99 ")
print(a)
b = validate_username("  Python_Dev ")
print(b)
```

Question 3: The Messy Database Cleaner
Task: You have received a raw string of names from a database exported as a semicolon-separated list. The data is messy, full of random spaces, and has inconsistent casing. Write a function to clean this string so that every name is properly Title Cased, extra spaces are removed, and they are joined together into a single string separated by a comma and a space.

Input:

raw_names = "  john doe ;   jAnE sMiTh;  ALAN tUrInG "
Expected Output:

"John Doe, Jane Smith, Alan Turing"


```python
def clean_names(raw_names):
    names = raw_names.split(";")

    cleaned_names = []

    for name in names:
        name = name.strip()
        name = name.title()
        cleaned_names.append(name)

    return ", ".join(cleaned_names)

clean_names ("  john doe ;   jAnE sMiTh;  ALAN tUrInG ")

```

Question 4: The Multiple Space Eliminator

Task: A user has accidentally typed a sentence with multiple random spaces between words, and the casing is entirely lowercase. Write a code that removes all the extra spaces so there is exactly one space between each word, and capitalize only the very first letter of the entire sentence. (Do not use the re module; solve it strictly using string functions.)


Input:

text = "   python    is   a    very   powerful  language   "
Expected Output:

"Python is a very powerful language"

```python
def clean_sentence(text):
    words = text.split()
    sentence = " ".join(words)
    return sentence.capitalize()

clean_sentence("   python    is   a    very   powerful  language   ")
```

Question 5: The Palindrome Sentence Checker


Task: Determine if a given sentence is a palindrome. To do this correctly using only string methods, remove all spaces and non-alphanumeric characters, convert everything to lowercase, and check if it reads the same forwards and backwards.


Input 1:

sentence = "Was it a car or a cat I saw?"


Expected Output 1:

True


Input 2:

sentence = "Python is fun"


Expected Output 2:

False

```python
def isPalindrome(sentence: str) -> bool:
   
    cleaned = "".join(char.lower() for char in sentence if char.isalnum())
    return cleaned == cleaned[::-1]

print(isPalindrome("Was it a car or a cat I saw?")) 
print(isPalindrome("Python is fun"))
```
