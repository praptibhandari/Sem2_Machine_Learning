```python
import pandas as pd
import numpy as np

df = pd.read_csv('../Data/tip.csv')
t = pd.read_csv('../Data/Titanic-Dataset.csv')

print(df.shape, t.shape)


# =========================
# Part A — tips.csv
# =========================

# A1
df['tip_percent'] = (df['tip'] / df['total_bill'] * 100).round(1)

print(df[['total_bill', 'tip', 'tip_percent']].head(5))


# A2
print(df.iloc[9, 0])
print(df.iloc[-3:, :2])


# A3
print(df.loc[20:25, 'sex':'size'])



# A4
print((~(df['time'] == 'Dinner')).sum())

print(((df['size'] >= 3) & (df['tip'] > 4)).sum())


# A5
df['service'] = 'normal'

df.loc[df['tip_percent'] > 20, 'service'] = 'generous'

print(df['service'].value_counts())


# =========================
# Part B — Titanic
# =========================

# B1
print(t['Embarked'].value_counts())


# B2
women_third = t[
    (t['Sex'] == 'female') &
    (t['Pclass'] == 3)
]

print(len(women_third))

print((women_third['Survived'] == 1).sum())


# B3
expensive = t[t['Fare'] > 200]

print(expensive[['Name', 'Age', 'Fare']])

print(len(expensive))


# B4
first_class = t[t['Pclass'] == 1]
third_class = t[t['Pclass'] == 3]

first_survival = first_class['Survived'].mean() * 100
third_survival = third_class['Survived'].mean() * 100

print(f"First class: {first_survival:.2f}%")
print(f"Third class: {third_survival:.2f}%")


# B5
t['age_group'] = 'adult'

t.loc[t['Age'] < 18, 'age_group'] = 'child'

print(t['age_group'].value_counts())
```
