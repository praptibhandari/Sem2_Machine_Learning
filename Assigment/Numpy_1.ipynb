Q15. Given the marks array below - 
(a) extract all marks greater than 60;

 (b) extract marks between 50 and 80 using two conditions with & (remember the parentheses!).

marks = np.array([56, 91, 44, 77, 62, 85, 39, 70])

```python
marks = np.array([56, 91, 44, 77, 62, 85, 39, 70])

# A
print(marks[marks > 60])

# B
print(marks[(marks >= 50) & (marks <= 80)])
```

| Concept             | Syntax                         |
| ------------------- | ------------------------------ |
| Create range        | `np.arange()`                  |
| Equal spacing       | `np.linspace()`                |
| Identity matrix     | `np.eye()`                     |
| Fill matrix         | `np.full()`                    |
| Shape               | `arr.shape`                    |
| Number of elements  | `arr.size`                     |
| Dimensions          | `arr.ndim`                     |
| Data type           | `arr.dtype`                    |
| Change dtype        | `arr.astype(int)`              |
| Normal indexing     | `arr[2]`                       |
| Last element        | `arr[-1]`                      |
| Slicing             | `arr[start:stop:step]`         |
| Reverse             | `arr[::-1]`                    |
| Boolean filtering   | `arr[arr > 60]`                |
| Multiple conditions | `arr[(arr > 50) & (arr < 80)]` |
